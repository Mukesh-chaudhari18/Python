# LOOPS

### What are Loops?
Loops are control structures that allow you to execute a block of code repeatedly. They are essential for automating repetitive tasks and processing collections of data. Python provides two main types of loops: `while` loops and `for` loops.

### Why use Loops?
*   **Efficiency:** Avoid writing the same code multiple times.
*   **Automation:** Perform tasks automatically until a condition is met.
*   **Iteration:** Go through each item in a list, string, or other collection.

## While Loop:

A `while` loop repeatedly executes a block of code as long as a specified condition is true. It's useful when you don't know in advance how many times you need to loop.

**Key Components of a `while` loop:**
1.  **Initialization:** Setting an initial value for a counter or a variable before the loop starts (e.g., `i = 1`).
2.  **Condition:** An expression that is evaluated before each iteration. If it's `True`, the loop continues; if `False`, the loop terminates (e.g., `i < 6`).
3.  **Increment/Decrement:** A statement inside the loop that changes the counter or variable, moving it towards the condition becoming false. This prevents an infinite loop (e.g., `i = i + 1` or `i += 1`).

```python
# Simple while loop
i=1 # initialization
while i<6: # condition
    print('While condition is true',i)
    i=i+1 # increment / decrement
```

This example shows a basic `while` loop. The loop prints the value of `i` as long as `i` is less than 6. After each print, `i` is increased by 1 (`i=i+1`). When `i` becomes 6, the condition `i < 6` becomes false, and the loop stops.

```python
i=1
while i<6:
    print('While condition is true',i)
    i+=1
print(i)
```

This is another example of a simple `while` loop, demonstrating a more compact way to increment `i` using `i += 1`. After the loop finishes, the final value of `i` (which made the condition false) is printed.

```python
# While loop with break.
#It is used to stop loop even if while condition is true
i=1
while i<6:
    print(i)
    if i==4:
        break
    i+=1
```

### `break` statement
The `break` statement is used to *immediately exit* a loop (both `while` and `for`) even if the loop's condition is still true or the iterable has more items. In this example, the loop stops as soon as `i` becomes 3, even though the `while i < 6` condition is still true.

```python
# While loop with continue statement.
i=0
while i<6:
    i+=1
    if i==5:
        pass

    print(i)
```

### `continue` statement
The `continue` statement is used to *skip the rest of the current iteration* of a loop and move to the next iteration. The loop does not terminate; it just bypasses the remaining code within the current loop body for that specific iteration. Here, when `i` is 3, the `print(i)` statement is skipped, and the loop continues with `i=4`.

```python
# While loop with else statement.
i=3
while i <6:
    i+=1
    print(i)
 #i=i+1
else:
    print('While loop condition is false')
```

### `else` with `while` loops
A `while` loop can also have an optional `else` block. The `else` block is executed *only if the loop completes normally* (i.e., the condition becomes false). It is NOT executed if the loop is terminated by a `break` statement. In this case, the loop completes when `i` becomes 6, and then the 'While loop condition is false' message is printed.

# FOR LOOP


### What are `for` Loops?
A `for` loop is used to iterate over a sequence (like a list, tuple, string, or range) or other iterable objects. It executes a block of code for *each item* in the sequence. It's typically used when you know in advance how many times you want to loop or when you need to process every item in a collection.

```python
l=[1,2,3,4,5,6,7,8,9,0,12,34,55]
for i in l:
    print(i)
```

This example shows iterating through a list `l`. The `for` loop assigns each element of the list to the variable `c` in successive iterations, and then `c` is printed.

```python
fruits=['Apple','Banana','Mango','Watermelon']
for i in fruits:
    print(i)
```

Similar to the previous example, this `for` loop iterates over a list of strings (fruits) and prints each one.

When you iterate over a string using a `for` loop, each character of the string is processed one by one.

### `range()` function
The `range()` function generates a sequence of numbers. It's often used with `for` loops to iterate a specific number of times.
*   `range(stop)`: Generates numbers from 0 up to (but not including) `stop`.
*   `range(start, stop)`: Generates numbers from `start` up to (but not including) `stop`.
*   `range(start, stop, step)`: Generates numbers from `start` up to (but not including) `stop`, incrementing by `step` each time.

This example uses `range(1, 10)` which generates numbers from 1 to 9.

```python
# odd values 
for i in range(0,101):
    if i %2 != 0:
        print(i)
```

This example uses `range(0, 20, 2)` to generate even numbers from 0 up to (but not including) 20, with a step of 2.

### Nested Loops
A nested loop is a loop inside another loop. The inner loop completes all its iterations for each single iteration of the outer loop. This is useful for tasks like processing items in a 2D grid or combining elements from multiple lists.

In the following example, for each subject in `list1`, the inner loop will iterate through all the numbers in `list2`.

```python
for i in list1: # Outer loop
    for z in list2: # inner loop
        print(i,z)
```

This code demonstrates a nested loop where every item from `list1` is paired with every item from `list2`.

```python
a=['red','blue','green']
b=['Apple','Banana','Orange']
for i in a: ## outer loop
    for h in b: ## inner loop
        print(i,h)
```

Another example of nested loops, showing how each color from `a` is paired with each fruit from `b`.

### Lambda Functions (Anonymous Functions)

Lambda functions are small, anonymous functions defined with the `lambda` keyword. They are called 'anonymous' because they don't have a formal `def` name. They are typically used for short, one-time operations where defining a full function with `def` would be overkill.

2.  **Single Expression:** Lambda functions can take any number of arguments but must have only one expression. The result of this expression is implicitly returned.
3.  **Short-lived:** We use lambda functions when we require a nameless function for a short period of time, often as arguments to higher-order functions like `map()`, `filter()`, and `reduce()`.

## Filter Function



### `filter()` Function

The `filter()` function constructs an iterator from elements of an iterable for which a function returns true. It offers an elegant way to filter out all the elements of an iterable (like a list) for which the function returns `False`.

**Syntax:** `filter(function, iterable)`
*   `function`: A function that tests if each element of an iterable returns true or false. Often, a `lambda` function is used here.
*   `iterable`: The sequence to be filtered (e.g., a list, tuple, string).

```python
# a list contains both even and odd numbers.
seq = [0, 1, 2, 3, 5, 8, 13,34,2,3,4,2,3,5,6]
```

```python
# result contains odd numbers of the list
result = filter(lambda i: i % 2 != 0, range(0,100))
print(list(result))
```

```python
# result contains even numbers of the list
result = filter(lambda y: y % 2 == 0, range(0,20))
print(list(result))
```

This example uses `filter()` with lambda functions to separate odd and even numbers from a list.
*   `lambda x: x % 2 != 0`: This lambda checks if a number is odd.
*   `lambda x: x % 2 == 0`: This lambda checks if a number is even.

The `filter()` function returns a filter object, which is then converted to a list using `list()` to display the results.

## Map Function

### `map()` Function

The `map()` function applies a given `function` to each item of an `iterable` (or multiple iterables) and returns an iterator of the results. It's useful when you want to transform every item in a sequence.

**Syntax:** `map(function, iterable, ...)`
*   `function`: The function to apply to each item.
*   `iterable`: One or more sequences whose items will be passed to the function.

```python
# Return double of n
def addition(n):
    return n + n

# We double all numbers using map()
numbers = (1, 2, 3, 4)
result = map(addition, numbers)
print(list(result))
```

This example defines a function `addition` that doubles a number. Then, `map(addition, numbers)` applies this `addition` function to each number in the `numbers` tuple, doubling each one. The result is then converted to a list.

```python
# Double all numbers using map and lambda
numbers = [1, 2, 3, 4,3,4,5,6]
result = map(lambda x: x + x, numbers)
print(list(result))
```

This example achieves the same doubling effect as the previous one but uses a `lambda` function directly within `map()`, making the code more concise. `lambda x: x + x` is the function applied to each `x` in `numbers`.

```python
print(list(result))
```

```python
# Square values of a list 

l1 = [ 23,54,54,24,12,33,22,11,10,4,3]
sq = map(lambda x : x ** 2 ,l1)
```

```python
# Add two lists using map and lambda

numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]

result = map(lambda x, y: x + y, numbers1, numbers2)
print(list(result))
```

The `map()` function can take multiple iterables. Here, `lambda x, y: x + y` takes two arguments, and `map()` pairs corresponding elements from `numbers1` and `numbers2` to apply the lambda function, effectively adding them element-wise.

```python
# List of strings
l = ['sat', 'bat', 'cat', 'mat']

# map() can listify the list of strings individually
test = list(map(list, l))
print(test)
```

This example shows a more advanced use of `map()`. It applies the built-in `list()` constructor to each string in `l`. This effectively converts each string into a list of its individual characters.

## Reduce Function

### `reduce()` Function

The `reduce()` function is part of the `functools` module. It applies a function of two arguments cumulatively to the items of an iterable, from left to right, so as to reduce the iterable to a single value.

**Syntax:** `functools.reduce(function, iterable[, initializer])`
*   `function`: The function to apply, taking two arguments.
*   `iterable`: The sequence to process.
*   `initializer` (optional): An initial value placed before the items of the iterable in the calculation.

```python
# importing functools for reduce()
import functools

# initializing list
lis = [1,4, 3, 5, 6, 2]

# using reduce to compute sum of list
print("The sum of the list elements is : ", end="")
print(functools.reduce(lambda a, b: a+b, lis))

```

```python
sum1 = functools.reduce(lambda a , b : a * b , range(1,10))

```

```python
sum1
```

This example uses `reduce()` to calculate the sum of all elements in the `lis` list. `lambda a, b: a + b` is applied repeatedly:
1.  `1 + 3 = 4`
2.  `4 + 5 = 9`
3.  `9 + 6 = 15`
4.  `15 + 2 = 17`
The final result is `17`.

```python
# using reduce to compute maximum element from list
print("The maximum element of the list is : ", end="")
print(functools.reduce(lambda a, b: a if a > b else b, lis))
```

Here, `reduce()` is used to find the maximum element in the list. `lambda a, b: a if a > b else b` compares two elements at a time and returns the larger one, progressively finding the overall maximum.
1.  `max(1, 3) = 3`
2.  `max(3, 5) = 5`
3.  `max(5, 6) = 6`
4.  `max(6, 2) = 6`
The final result is `6`.
